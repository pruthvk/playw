```python
import sys
import io

def filter_students(student_data: dict, threshold: float) -> dict:
    # Dictionary comprehension to calculate average and filter
    return {
        name: sum(scores) / len(scores) 
        for name, scores in student_data.items() 
        if (sum(scores) / len(scores)) > threshold
    }
# --- SIMULATE HACKEREARTH CUSTOM INPUT BOX ---
hacker_earth_input = """3
80
Alice 90, 95, 100
Bob 60, 70, 65
Charlie 80, 85, 90"""

sys.stdin = io.StringIO(hacker_earth_input)
# ---------------------------------------------
if __name__ == "__main__":
    try:
        # 1. Read first line: Number of students
        num_students_line = sys.stdin.readline().strip()
        if not num_students_line:
            sys.exit(0)
            
        num_students = int(num_students_line)
        
        # 2. Read second line: Threshold float
        threshold = float(sys.stdin.readline().strip())
        
        # 3. Read the next 'n' lines for student data
        student_data = {}
        for _ in range(num_students):
            line_parts = sys.stdin.readline().strip().split()
            name = line_parts[0]
            scores = [float(score) for score in line_parts[1:]]
            student_data[name] = scores
            
        # 4. Filter the students
        filtered_result = filter_students(student_data, threshold)
        
        # 5. Sort alphabetically by name and print each with 2 decimal places
        # sorted() on a dictionary sorts by keys (names) automatically
        for name in sorted(filtered_result.keys()):
            average_score = filtered_result[name]
            print(f"{name} {average_score:.2f}")

    except (ValueError, IndexError) as e:
        print(f"Error parsing input format: {e}", file=sys.stderr)
        
        
```

```python
# You are given a long string your task is to find the length of the longest upstring without repeating characters.

def longest_substring(s: str) -> int:
    seen = set()
    start = 0
    max_length = 0

    for end in range(len(s)):
        # While the current letter is a duplicate, shrink the window from the left
        while s[end] in seen:
            seen.remove(s[start])
            start += 1
            
        # Add the new letter and update the max length
        seen.add(s[end])
        max_length = max(max_length, end - start + 1)
        
    return max_length
```

```python
# You need to generate specific sequence. That alternate between zeros and consecutive positive positive integers. Given a number in create a string where each integer from 1 to Yen is preceded by zero. farming a pattern like 0102030405
def generate_sequence(n: int) -> str:
    result = ""  
    
    for i in range(1, n + 1):
        # Convert the number directly to a string (no % 10)
        digit_string = str(i)
        
        # Safely add "0" and the number to our result string
        result += "\t0" + digit_string 
        
    return result

# --- TEST ---
print(generate_sequence(5))   # Output: 0102030405
print(generate_sequence(12))  # Output: 010203040506070809010011012
print(generate_sequence(20))
print(generate_sequence(100))
print(generate_sequence(200))
print(generate_sequence(1200))
```

```python
# # Line 1: An integer $T$ representing the total number of test cases to run.

# The Test Case Block: Right after line 1, the text is divided into $T$ separate blocks. Every single block follows this strict 3-line sequence:

# First line of the case: An integer $N$ denoting the expected length of the strings for this specific run.

# Second line of the case: String $s$, containing lowercase English letters and wildcard ? characters.

# Third line of the case: String $t$, containing only lowercase English letters.

import sys
from collections import Counter

def solve():
    # Read the entire input line by line
    raw_lines = sys.stdin.read().splitlines()
    
    # Store clean, non-empty lines
    lines = []
    for line in raw_lines:
        cleaned = line.strip()
        if cleaned:
            lines.append(cleaned)
            
    if not lines:
        return

    # Line 1 is now safely captured as T because we aren't burning it anymore!
    num_test_cases = int(lines[0])
    
    # Start looking for data blocks at index 1
    line_idx = 1
    
    for _ in range(num_test_cases):
        # 1. Read N (String length)
        n = int(lines[line_idx])
        
        # 2. Read string s
        s = lines[line_idx + 1]
        
        # 3. Read string t
        t = lines[line_idx + 2]
        
        # Advance by 3 lines to step to the next block
        line_idx += 3
        
        # --- Core Match Logic ---
        if len(s) != n or len(t) != n:
            print("No")
            continue
            
        count_s = Counter(s)
        count_t = Counter(t)
        
        needed_replacements = 0
        
        for char, required_count in count_t.items():
            current_count = count_s[char]
            if current_count < required_count:
                needed_replacements += (required_count - current_count)
                
        if needed_replacements <= count_s['?']:
            print("Yes")
        else:
            print("No")

if __name__ == "__main__":
    solve()
```
